2024-10-17과 2025-11-06을 비교하며, 공급구조가 유지된 126개 250m 격자만 분석한다.

```python
# 프로젝트 루트(UROP) 자동 탐색 — 절대경로 없이 어느 컴퓨터에서도 동작
import sys
from pathlib import Path
UROP_ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'config.py').is_file() and (p / 'code').is_dir())
for _d in [UROP_ROOT, *sorted(d for d in (UROP_ROOT / 'code').rglob('*') if d.is_dir() and d.name != '__pycache__')]:
    if str(_d) not in sys.path:
        sys.path.append(str(_d))

from pathlib import Path
import warnings
import geopandas as gpd
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from matplotlib.colors import TwoSlopeNorm
from matplotlib.animation import FuncAnimation, PillowWriter
from IPython.display import display, Image
warnings.filterwarnings('ignore')

PROJECT = UROP_ROOT
ROOT = UROP_ROOT
PROCESSED = PROJECT / 'data' / 'processed' / 'gtx_a'
FIGURES = PROJECT / 'results' / 'eda' / 'figures_011'
FIGURES.mkdir(parents=True, exist_ok=True)
STABLE_PATH = PROCESSED / '16_새데이터_공급구조유지_250m격자.csv'
MODEL_PATH = PROCESSED / '18_새데이터_회귀분석_격자통합데이터.csv'
GRID_PATH = next((ROOT / 'data' / 'raw' / 'common' / 'grid' / '빈격자(250m)').glob('*.shp'))
ADMIN_PATH = ROOT / 'data' / 'raw' / 'common' / 'api_cache_gtx_a_routes' / 'HangJeongDong_ver20260701.geojson'
CARD_DIR = ROOT / 'data' / 'raw' / 'gtx_a' / 'transport_card'
RAW24 = CARD_DIR / 'gtx_a_transport_card_20241017_raw_with_coords_api_filled.csv'
RAW25 = CARD_DIR / 'gtx_a_transport_card_20251106_raw_with_coords.csv'

font_candidates = [
    *(UROP_ROOT / 'code').rglob('12-8_A3_쌍대비교_순위모형.py'),
    Path(r'C:\Windows\Fonts\malgun.ttf'),
    Path('/System/Library/Fonts/AppleSDGothicNeo.ttc'),
    Path('/Library/Fonts/NanumGothic.ttf'),
]
font_path = next((path for path in font_candidates if path.exists()), None)
if font_path is not None:
    fm.fontManager.addfont(str(font_path))
    plt.rcParams['font.family'] = fm.FontProperties(fname=str(font_path)).get_name()
plt.rcParams['axes.unicode_minus'] = False

stable = pd.read_csv(STABLE_PATH, encoding='utf-8-sig')
model = pd.read_csv(MODEL_PATH, encoding='utf-8-sig')
stable['셀 ID'] = pd.to_numeric(stable['셀 ID'], errors='coerce').astype('Int64')
model['셀 ID'] = pd.to_numeric(model['셀 ID'], errors='coerce').astype('Int64')
model['변화율(%)'] = np.where(model['board_2024_all'] > 0, model['change_all'] / model['board_2024_all'] * 100, np.nan)
assert len(stable) == 126 and len(model) == 126

admin = gpd.read_file(ADMIN_PATH)
grid_crs = gpd.read_file(GRID_PATH, rows=1).crs
admin = admin.to_crs(grid_crs)
admin['sgg_code'] = admin['sgg'].astype(str).str.extract(r'([0-9]+)')[0].str.zfill(5)
city = admin[admin['sgg_code'].str.startswith(('4128', '4148'))]
boundary = city.geometry.union_all() if hasattr(city.geometry, 'union_all') else city.geometry.unary_union
grid = gpd.read_file(GRID_PATH, bbox=boundary.bounds)
grid = grid[grid.geometry.intersects(boundary)].copy()
grid['_key'] = grid.geometry.apply(lambda x: x.wkb if x is not None else None)
grid = grid.drop_duplicates('_key').drop(columns='_key').reset_index(drop=True)
grid['셀 ID'] = np.arange(1, len(grid) + 1)
plot_grid = grid[['셀 ID', 'geometry']].merge(model, on='셀 ID', how='inner', validate='one_to_one')
boundary_gdf = gpd.GeoDataFrame(geometry=[boundary], crs=grid.crs)

stations = gpd.GeoDataFrame(
    {'역명': ['운정중앙역', '킨텍스역', '대곡역']},
    geometry=gpd.points_from_xy([126.728330, 126.748060, 126.811024], [37.716670, 37.665000, 37.631626]),
    crs='EPSG:4326'
).to_crs(grid.crs)
station_join = gpd.sjoin(stations[['역명', 'geometry']], grid[['셀 ID', 'geometry']], how='left', predicate='within')
station_cell_ids = set(station_join['셀 ID'].dropna().astype(int))
station_cells = grid[grid['셀 ID'].isin(station_cell_ids)]
print('분석 격자:', len(plot_grid), '| GTX-A역 포함 격자:', sorted(station_cell_ids))
display(model[['셀 ID', 'board_2024_all', 'board_2025_all', 'change_all', '변화율(%)']].head())
```

## 1. 격자별 승차건수 변화량 지도: 역 권역별 확대
126개 격자의 2025년 − 2024년 승차건수 변화량을 표시한다. 감소는 공통 사분위에 따른 붉은색, 동일은 회색, 증가는 파란색이다. 각 권역의 감소량 TOP 3를 노란 테두리로 표시하고, 숫자 없는 버전과 1·2·3 표시 버전을 함께 제시한다.


```python
# 역 권역별 승차건수 변화량 확대 지도: 권역별 2개 패널, 126개 공통 경계
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, TwoSlopeNorm
from matplotlib.patches import Patch

# 뒤의 기존 변화율 애니메이션 셀에서 사용하는 정규화 유지
limit = max(1, float(plot_grid['변화율(%)'].abs().quantile(0.95)))
norm = TwoSlopeNorm(vmin=-limit, vcenter=0, vmax=limit)

# 건수 기준 입력을 셀 ID로 연결한다. 기존 plot_grid는 변경하지 않는다.
from pathlib import Path
import math
count_static_root = next(p for p in [Path.cwd(), *Path.cwd().parents]
                         if (p / 'config.py').is_file() and (p / 'code').is_dir())
count_static_data = pd.read_csv(
    count_static_root / 'data/final/gtx_a/S1_126격자_건수기준_기초표.csv', encoding='utf-8-sig'
)
assert len(count_static_data) == 126 and count_static_data['셀 ID'].is_unique
assert set(plot_grid['셀 ID'].astype(int)) == set(count_static_data['셀 ID'].astype(int))
count_plot_grid = plot_grid[['셀 ID', 'geometry']].merge(count_static_data, on='셀 ID', validate='one_to_one')
assert count_plot_grid[['board_2024', 'board_2025', 'change']].notna().all().all()
np.testing.assert_array_equal(count_plot_grid['change'], count_plot_grid.board_2025-count_plot_grid.board_2024)

# 126개 공통 사분위 경계. 증가=파랑, 동일=회색으로 별도 표시한다.
count_static_edges = count_plot_grid['change'].quantile([.25, .50, .75]).to_numpy()
quartile_colors = ['#67000D', '#CB181D', '#FB6A4A', '#FEE0D2']
count_plot_grid['변화량 사분위 코드'] = np.searchsorted(count_static_edges, count_plot_grid['change'], side='left')
count_plot_grid['건수 지도색'] = [
    '#2166AC' if value > 0 else '#EEEEEE' if value == 0 else quartile_colors[int(code)]
    for value, code in zip(count_plot_grid['change'], count_plot_grid['변화량 사분위 코드'])
]
quartile_handles = []
count_static_bins = []
for index, color in enumerate(quartile_colors):
    lower = None if index == 0 else math.floor(count_static_edges[index-1])+1
    upper = min(math.floor(count_static_edges[index]), -1) if index < 3 else -1
    if lower is not None and lower > upper:
        continue
    label = (f'{upper:,}건 이하' if lower is None else
             f'{lower:,}건' if lower == upper else f'{lower:,}건 이상 {upper:,}건 이하')
    quartile_handles.append(Patch(facecolor=color, edgecolor='#333333', label=label))
    mask = count_plot_grid['change'].le(upper)
    if lower is not None: mask &= count_plot_grid['change'].ge(lower)
    count_static_bins.append({'변화량 범위': label, '격자 수': int(mask.sum())})
for value, color, label in [(0, '#EEEEEE', '0건 (동일)'), (1, '#2166AC', '+1건 이상 (증가)')]:
    quartile_handles.append(Patch(facecolor=color, edgecolor='#333333', label=label))
    mask = count_plot_grid['change'].eq(0) if value == 0 else count_plot_grid['change'].gt(0)
    count_static_bins.append({'변화량 범위': label, '격자 수': int(mask.sum())})
print('건수 기준 변화량 공통 사분위 경계:', tuple(count_static_edges), '건')
display(pd.DataFrame(count_static_bins))

station_order = ['운정중앙', '킨텍스', '대곡']
panel_titles = {'운정중앙': '운정중앙 권역', '킨텍스': '킨텍스 권역', '대곡': '대곡 권역'}

def draw_region_map(ax, region, top3_region, station_name, bounds, show_rank=False):
    xmin, ymin, xmax, ymax = bounds

    # 기존 확대지도 형식: 주변 전체 격자는 연한 회색, 분석격자만 변화량 색상
    background = grid.cx[xmin:xmax, ymin:ymax]
    background.plot(
        ax=ax, facecolor='#F2F2F2', edgecolor='#D2D2D2',
        linewidth=0.28, zorder=1,
    )
    region.plot(
        ax=ax, color=region['건수 지도색'], edgecolor='#222222',
        linewidth=1.15, alpha=0.92, zorder=3,
    )

    # 권역별 절대 감소량 TOP 3는 노란 테두리로 강조
    if len(top3_region):
        top3_region.boundary.plot(
            ax=ax, color='#FFD400', linewidth=3.6, zorder=6,
        )

    # 숫자 표시 버전에는 TOP 1~3을 각각 1·2·3으로만 표시
    if show_rank:
        for _, row in top3_region.iterrows():
            point = row.geometry.representative_point()
            ax.text(
                point.x, point.y, str(int(row['감소 순위'])),
                ha='center', va='center', fontsize=13, fontweight='bold',
                color='#111111', zorder=8,
                bbox=dict(
                    boxstyle='circle,pad=0.26', facecolor='#FFD400',
                    edgecolor='white', linewidth=1.2, alpha=0.96,
                ),
            )

    station_point = stations[
        stations['역명'].str.replace('역', '', regex=False) == station_name
    ]
    station_point.plot(
        ax=ax, color='#00A651', marker='*', markersize=220,
        edgecolor='white', linewidth=1.0, zorder=7,
    )
    boundary_gdf.boundary.plot(ax=ax, color='#777777', linewidth=0.8, zorder=2)

    ax.set_xlim(xmin, xmax)
    ax.set_ylim(ymin, ymax)
    ax.set_aspect('equal')
    ax.set_title(
        '순위 숫자 표시' if show_rank else '노란 테두리만 표시',
        fontsize=14, fontweight='bold', pad=10,
    )
    ax.set_axis_off()


for station_name in station_order:
    region = count_plot_grid[count_plot_grid['nearest_station'].astype(str) == station_name].copy()
    top3_region = region[region['change'] < 0].nsmallest(3, 'change').copy().reset_index(drop=True)
    top3_region['감소 순위'] = np.arange(1, len(top3_region) + 1)

    xmin, ymin, xmax, ymax = region.total_bounds
    pad_x = max((xmax - xmin) * 0.08, 350)
    pad_y = max((ymax - ymin) * 0.08, 350)
    bounds = (xmin-pad_x, ymin-pad_y, xmax+pad_x, ymax+pad_y)

    # 권역마다 동일한 범위의 지도 2개: 테두리 전용 / 숫자 표시
    fig, axes = plt.subplots(1, 2, figsize=(18, 8.5))
    draw_region_map(axes[0], region, top3_region, station_name, bounds, show_rank=False)
    draw_region_map(axes[1], region, top3_region, station_name, bounds, show_rank=True)

    fig.suptitle(
        f'{panel_titles[station_name]} 광역버스 승차건수 변화량',
        fontsize=19, fontweight='bold', y=0.98,
    )
    fig.legend(
        handles=quartile_handles, loc='lower center', ncol=3,
        frameon=True, framealpha=0.96, bbox_to_anchor=(0.5, 0.055), fontsize=10.5,
    )
    fig.text(
        0.5, 0.018,
        '노란 테두리: 해당 권역의 절대 감소량 TOP 3   숫자 1·2·3: 감소량 순위   ★ GTX-A역',
        ha='center', fontsize=10, color='#444444',
    )
    plt.tight_layout(rect=[0, 0.12, 1, 0.94])
    plt.show()

# 권역별 절대 감소량 TOP 3 중심좌표: 지도 검색창에는 "위도, 경도"를 그대로 입력
top3_rows = []
for station_name in station_order:
    part = (
        count_plot_grid[(count_plot_grid['nearest_station'].astype(str) == station_name) & (count_plot_grid['change'] < 0)]
        .nsmallest(3, 'change')
        .copy()
        .reset_index(drop=True)
    )
    part['감소 순위'] = np.arange(1, len(part) + 1)
    top3_rows.append(part)

top3_coordinates = pd.concat(top3_rows, ignore_index=True)[
    ['nearest_station', '감소 순위', '셀 ID', '중심점 위도', '중심점 경도',
     'board_2024', 'board_2025', 'change']
].copy()
top3_coordinates['지도 검색용 좌표'] = (
    top3_coordinates['중심점 위도'].map(lambda x: f'{x:.6f}') + ', ' +
    top3_coordinates['중심점 경도'].map(lambda x: f'{x:.6f}')
)
top3_coordinates = top3_coordinates.rename(columns={
    'nearest_station': '권역',
    'board_2024': '2024년 승차건수',
    'board_2025': '2025년 승차건수',
    'change': '변화량',
})
print('권역별 절대 감소량 TOP 3 좌표(총 9개)')
display(top3_coordinates.round({
    '중심점 위도': 6, '중심점 경도': 6,
    '2024년 승차건수': 0, '2025년 승차건수': 0, '변화량': 0,
}))
```

### 출근시간 승차건수 변화량 격자 지도: 권역별 확대

종일 정적 지도와 같은 방식으로 **출근시간 06:00~09:59:59의 126개 격자**를 표시한다. 이 셀은 앞선 셀이나 카카오 API 키 없이 단독 실행할 수 있다.

- 운정중앙·킨텍스·대곡 **권역마다 한 장씩**, 같은 범위의 두 패널(테두리만 / 1·2·3 표시)을 제시한다.
- 증가 파랑·동일 회색·감소 빨강 계열이다. **출근시간 전체 126개 공통 사분위**를 적용하며 종일 지도와 경계가 다를 수 있다.
- 노란 테두리는 **각 권역 출근시간 감소 TOP 3**다. 종일 TOP 3와 다를 수 있으며, 동률이면 셀 ID 순서로 정한다.
- 마지막 표에 TOP 3 전후 건수·변화량·좌표를 표시한다. 2024년 0건 격자도 변화량이 정의되므로 포함한다.
- 기존 정적 지도·카카오 지도 데이터는 유지하며 PNG·CSV 파일은 별도 저장하지 않는다.


```python
# 출근시간 카카오맵: 06:00~09:59:59 · 126개 격자 전체 · 건수 변화량
from pathlib import Path
import re
import numpy as np
import pandas as pd
import geopandas as gpd

am_static_root = next((p for p in [Path.cwd(), *Path.cwd().parents]
                if (p/'config.py').is_file() and (p/'code').is_dir()), None)
if am_static_root is None:
    raise FileNotFoundError('UROP 폴더 또는 그 하위 폴더를 작업 디렉터리로 열어 주세요.')
am_static_base = pd.read_csv(am_static_root/'data/final/gtx_a/S1_126격자_건수기준_기초표.csv', encoding='utf-8-sig')
am_static_cells = pd.read_csv(am_static_root/'data/processed/gtx_a/16_새데이터_공급구조유지_250m격자.csv',dtype=str,encoding='utf-8-sig')
assert len(am_static_base)==126 and am_static_base['셀 ID'].is_unique
assert set(am_static_base['셀 ID'])==set(am_static_cells['셀 ID'].astype(int))
am_static_counts = am_static_base[['셀 ID','nearest_station']].copy().set_index('셀 ID')
for am_static_year,am_static_filename,am_static_date in [
    (2024,'gtx_a_transport_card_20241017_raw_with_coords_api_filled.csv','20241017'),
    (2025,'gtx_a_transport_card_20251106_raw_with_coords.csv','20251106')]:
    am_static_card = pd.read_csv(am_static_root/'data/raw/gtx_a/transport_card'/am_static_filename,dtype=str,encoding='utf-8-sig')
    assert am_static_card['query_date'].eq(am_static_date).all() and am_static_card['opr_ymd'].eq(am_static_date).all()
    am_static_lookup = pd.DataFrame([(int(row['셀 ID']),sid) for _,row in am_static_cells.iterrows()
                              for sid in re.findall(r'\d{6,}',row[f'{am_static_year}년 정류장 ID'])],
                             columns=['셀 ID','승차정류장ID'])
    assert not am_static_lookup['승차정류장ID'].duplicated().any()
    am_static_selected = am_static_card.merge(am_static_lookup,on='승차정류장ID',how='inner',validate='many_to_one')
    np.testing.assert_array_equal(
        am_static_selected.groupby('셀 ID').size().reindex(am_static_counts.index,fill_value=0),
        am_static_base.set_index('셀 ID').loc[am_static_counts.index,f'board_{am_static_year}'])
    am_static_times = pd.to_datetime(am_static_selected['ride_dt'],format='%Y%m%d%H%M%S',errors='raise')
    # 9시대 포함, 10시대 제외. utztn_nope 합산이나 중복 제거는 하지 않는다.
    am_static_selected = am_static_selected.loc[am_static_times.dt.hour.ge(6) & am_static_times.dt.hour.lt(10)]
    am_static_counts[f'board_{am_static_year}'] = am_static_selected.groupby('셀 ID').size().reindex(am_static_counts.index,fill_value=0)
am_static_counts['change'] = am_static_counts['board_2025']-am_static_counts['board_2024']
am_static_reference = pd.read_csv(am_static_root/'results/analysis/feeder_bus/tables/H11a_격자별_공급구조_GTX역행.csv',encoding='utf-8-sig').set_index('셀 ID')
np.testing.assert_array_equal(am_static_counts['change'],am_static_reference.loc[am_static_counts.index,'Δ광역 출근'])

# 기존 원본과 같은 행정계 필터·중복 제거·생성 순서로 셀 ID를 재현한다.
am_static_admin_path = am_static_root/'data/raw/common/api_cache_gtx_a_routes/HangJeongDong_ver20260701.geojson'
am_static_grid_path = next((am_static_root/'data/raw/common/grid/빈격자(250m)').glob('*.shp'))
am_static_grid_crs = gpd.read_file(am_static_grid_path,rows=1).crs
am_static_admin = gpd.read_file(am_static_admin_path).to_crs(am_static_grid_crs)
am_static_admin['sgg_code'] = am_static_admin['sgg'].astype(str).str.extract(r'([0-9]+)')[0].str.zfill(5)
am_static_city = am_static_admin[am_static_admin['sgg_code'].str.startswith(('4128','4148'))]
am_static_boundary = am_static_city.geometry.union_all() if hasattr(am_static_city.geometry,'union_all') else am_static_city.geometry.unary_union
am_static_geometry = gpd.read_file(am_static_grid_path,bbox=am_static_boundary.bounds)
am_static_geometry = am_static_geometry[am_static_geometry.geometry.intersects(am_static_boundary)].copy()
am_static_geometry['_key'] = am_static_geometry.geometry.apply(lambda g:g.wkb if g is not None else None)
am_static_geometry = am_static_geometry.drop_duplicates('_key').drop(columns='_key').reset_index(drop=True)
am_static_geometry['셀 ID'] = np.arange(1,len(am_static_geometry)+1)
am_static_map_grid = am_static_geometry[['셀 ID','geometry']].merge(am_static_counts.reset_index(),on='셀 ID',how='inner',validate='one_to_one')
assert len(am_static_map_grid)==126 and am_static_map_grid['셀 ID'].is_unique
assert am_static_map_grid[['board_2024','board_2025','change']].notna().all().all()
am_static_map_edges = tuple(am_static_map_grid['change'].quantile([.25,.50,.75]))
am_static_map_grid['변화량 사분위 코드'] = np.searchsorted(am_static_map_edges,am_static_map_grid['change'].to_numpy(),side='left')
# 종일 페이지는 원래 S1 값으로 유지하고 출근 페이지와 별도 URL로 제공한다.
am_static_day_grid = am_static_geometry[['셀 ID','geometry']].merge(
    am_static_base[['셀 ID','nearest_station','board_2024','board_2025','change']],on='셀 ID',how='inner',validate='one_to_one')
am_static_day_edges = tuple(am_static_day_grid['change'].quantile([.25,.50,.75]))
am_static_day_grid['변화량 사분위 코드'] = np.searchsorted(am_static_day_edges,am_static_day_grid['change'].to_numpy(),side='left')
print('출근시간: 06:00~09:59:59 · 126개 격자 전체')
print('2024년:',int(am_static_map_grid['board_2024'].sum()),'건 → 2025년:',int(am_static_map_grid['board_2025'].sum()),'건')
print('변화량:',int(am_static_map_grid['change'].sum()),'건 · 사분위 경계:',am_static_map_edges)
print('감소:',int(am_static_map_grid['change'].lt(0).sum()),'동일:',int(am_static_map_grid['change'].eq(0).sum()),'증가:',int(am_static_map_grid['change'].gt(0).sum()))


# 정적 지도: 권역별 전체 위치 확대, 노란 테두리 / 숫자 표시 두 버전
import math
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from matplotlib.patches import Patch
from IPython.display import display

for am_static_font in [Path('/System/Library/Fonts/AppleSDGothicNeo.ttc'),
                       Path(r'C:\Windows\Fonts\malgun.ttf'),Path('/Library/Fonts/NanumGothic.ttf')]:
    if am_static_font.is_file():
        fm.fontManager.addfont(str(am_static_font))
        plt.rcParams['font.family']=fm.FontProperties(fname=str(am_static_font)).get_name()
        break
plt.rcParams['axes.unicode_minus']=False
am_static_colors=['#67000D','#CB181D','#FB6A4A','#FEE0D2']
am_static_map_grid['건수 지도색']=[
    '#2166AC' if v>0 else '#EEEEEE' if v==0 else am_static_colors[int(c)]
    for v,c in zip(am_static_map_grid['change'],am_static_map_grid['변화량 사분위 코드'])]
am_static_legend=[];am_static_bins=[]
for i,color in enumerate(am_static_colors):
    lower=None if i==0 else math.floor(am_static_map_edges[i-1])+1
    upper=min(math.floor(am_static_map_edges[i]),-1) if i<3 else -1
    if lower is not None and lower>upper:
        continue
    label=(f'{upper:,}건 이하' if lower is None else
           f'{lower:,}건' if lower==upper else f'{lower:,}건 이상 {upper:,}건 이하')
    am_static_legend.append(Patch(facecolor=color,edgecolor='#333333',label=label))
    mask=am_static_map_grid['change'].le(upper)
    if lower is not None:mask &= am_static_map_grid['change'].ge(lower)
    am_static_bins.append({'출근시간 변화량 범위':label,'격자 수':int(mask.sum())})
for label,color,mask in [
    ('0건 (동일)','#EEEEEE',am_static_map_grid['change'].eq(0)),
    ('+1건 이상 (증가)','#2166AC',am_static_map_grid['change'].gt(0))]:
    am_static_legend.append(Patch(facecolor=color,edgecolor='#333333',label=label))
    am_static_bins.append({'출근시간 변화량 범위':label,'격자 수':int(mask.sum())})
assert sum(v['격자 수'] for v in am_static_bins)==126
display(pd.DataFrame(am_static_bins))
am_static_boundary_gdf=gpd.GeoDataFrame(geometry=[am_static_boundary],crs=am_static_grid_crs)
am_static_stations=gpd.GeoDataFrame(
    {'역명':['운정중앙','킨텍스','대곡']},
    geometry=gpd.points_from_xy([126.728330,126.748060,126.811024],[37.716670,37.665000,37.631626]),
    crs='EPSG:4326').to_crs(am_static_grid_crs)

def draw_morning_static_011(ax,region,top,station,bounds,show_rank=False):
    x0,y0,x1,y1=bounds
    background=am_static_geometry.cx[x0:x1,y0:y1]
    background.plot(ax=ax,facecolor='#F2F2F2',edgecolor='#D2D2D2',linewidth=.28,zorder=1)
    am_static_boundary_gdf.boundary.plot(ax=ax,color='#777777',linewidth=.8,zorder=2)
    region.plot(ax=ax,color=region['건수 지도색'],edgecolor='#222222',linewidth=1.15,alpha=.92,zorder=3)
    if len(top):top.boundary.plot(ax=ax,color='#FFD400',linewidth=3.6,zorder=6)
    if show_rank:
        for _,row in top.iterrows():
            point=row.geometry.representative_point()
            ax.text(point.x,point.y,str(int(row['감소 순위'])),ha='center',va='center',
                    fontsize=13,fontweight='bold',color='#111111',zorder=8,
                    bbox=dict(boxstyle='circle,pad=.26',facecolor='#FFD400',edgecolor='white',linewidth=1.2,alpha=.96))
    am_static_stations.loc[am_static_stations['역명'].eq(station)].plot(
        ax=ax,color='#00A651',marker='*',markersize=220,edgecolor='white',linewidth=1,zorder=7)
    ax.set_xlim(x0,x1);ax.set_ylim(y0,y1);ax.set_aspect('equal');ax.set_axis_off()
    ax.set_title('순위 숫자 표시' if show_rank else '노란 테두리만 표시',fontsize=14,fontweight='bold',pad=10)

am_static_top_rows=[]
for am_static_station in ['운정중앙','킨텍스','대곡']:
    am_static_region=am_static_map_grid.loc[am_static_map_grid['nearest_station'].eq(am_static_station)].copy()
    am_static_top=am_static_region.loc[am_static_region['change'].lt(0)].sort_values(['change','셀 ID'],kind='stable').head(3).copy()
    am_static_top['감소 순위']=np.arange(1,len(am_static_top)+1)
    am_static_top_rows.append(am_static_top)
    x0,y0,x1,y1=am_static_region.total_bounds
    px=max((x1-x0)*.08,350);py=max((y1-y0)*.08,350)
    bounds=(x0-px,y0-py,x1+px,y1+py)
    fig,axes=plt.subplots(1,2,figsize=(18,8.5))
    draw_morning_static_011(axes[0],am_static_region,am_static_top,am_static_station,bounds,False)
    draw_morning_static_011(axes[1],am_static_region,am_static_top,am_static_station,bounds,True)
    fig.suptitle(f'{am_static_station} 권역 출근시간 광역버스 승차건수 변화량',fontsize=19,fontweight='bold',y=.98)
    fig.legend(handles=am_static_legend,loc='lower center',ncol=3,frameon=True,framealpha=.96,
               bbox_to_anchor=(.5,.055),fontsize=10.5)
    fig.text(.5,.018,'06:00~09:59:59 · 노란 테두리: 권역 내 감소 TOP 3 · 숫자 1·2·3: 감소 순위 · ★ GTX-A역',
             ha='center',fontsize=10,color='#444444')
    plt.tight_layout(rect=[0,.12,1,.94]);plt.show()

# 지도에 표시한 감소 TOP 3와 검색용 좌표(실제 중심점)를 별도로 제시한다.
am_static_top3=pd.concat(am_static_top_rows,ignore_index=True).merge(
    am_static_base[['셀 ID','중심점 위도','중심점 경도']],on='셀 ID',validate='one_to_one')
am_static_top3['지도 검색용 좌표']=am_static_top3['중심점 위도'].map(lambda v:f'{v:.6f}')+', '+am_static_top3['중심점 경도'].map(lambda v:f'{v:.6f}')
am_static_top3=am_static_top3[['nearest_station','감소 순위','셀 ID','중심점 위도','중심점 경도',
    'board_2024','board_2025','change','지도 검색용 좌표']].rename(columns={
    'nearest_station':'권역','board_2024':'2024년 출근 승차건수','board_2025':'2025년 출근 승차건수','change':'출근 변화량(건)'})
print('출근시간 감소 TOP 3 좌표: 권역별 3개, 총 9개')
display(am_static_top3.round({'중심점 위도':6,'중심점 경도':6}))
print('화면 출력 완료 · 파일 저장 없음 · 기존 종일 지도 변수 유지')
```

- 2024년 출근 0건인 13개도 건수와 변화량을 계산할 수 있으므로 지도에 포함한다. 회귀분석의 113개 표본과 구분한다.
- 같은 localhost:8765에서 종일·출근 지도를 구분해 제공한다. 앞선 종일 지도와 다른 변수를 사용하므로 기존 정적 지도와 종일 집계값은 변경하지 않는다.

```python
# 출근시간 카카오맵: 06:00~09:59:59 · 126개 격자 전체 · 건수 변화량
from pathlib import Path
import re
import os
import sys
import types
import getpass
import numpy as np
import pandas as pd
import geopandas as gpd

am_root = next((p for p in [Path.cwd(), *Path.cwd().parents]
                if (p/'config.py').is_file() and (p/'code').is_dir()), None)
if am_root is None:
    raise FileNotFoundError('UROP 폴더 또는 그 하위 폴더를 작업 디렉터리로 열어 주세요.')
am_base = pd.read_csv(am_root/'data/final/gtx_a/S1_126격자_건수기준_기초표.csv', encoding='utf-8-sig')
am_cells = pd.read_csv(am_root/'data/processed/gtx_a/16_새데이터_공급구조유지_250m격자.csv',dtype=str,encoding='utf-8-sig')
assert len(am_base)==126 and am_base['셀 ID'].is_unique
assert set(am_base['셀 ID'])==set(am_cells['셀 ID'].astype(int))
am_counts = am_base[['셀 ID','nearest_station']].copy().set_index('셀 ID')
for am_year,am_filename,am_date in [
    (2024,'gtx_a_transport_card_20241017_raw_with_coords_api_filled.csv','20241017'),
    (2025,'gtx_a_transport_card_20251106_raw_with_coords.csv','20251106')]:
    am_card = pd.read_csv(am_root/'data/raw/gtx_a/transport_card'/am_filename,dtype=str,encoding='utf-8-sig')
    assert am_card['query_date'].eq(am_date).all() and am_card['opr_ymd'].eq(am_date).all()
    am_lookup = pd.DataFrame([(int(row['셀 ID']),sid) for _,row in am_cells.iterrows()
                              for sid in re.findall(r'\d{6,}',row[f'{am_year}년 정류장 ID'])],
                             columns=['셀 ID','승차정류장ID'])
    assert not am_lookup['승차정류장ID'].duplicated().any()
    am_selected = am_card.merge(am_lookup,on='승차정류장ID',how='inner',validate='many_to_one')
    np.testing.assert_array_equal(
        am_selected.groupby('셀 ID').size().reindex(am_counts.index,fill_value=0),
        am_base.set_index('셀 ID').loc[am_counts.index,f'board_{am_year}'])
    am_times = pd.to_datetime(am_selected['ride_dt'],format='%Y%m%d%H%M%S',errors='raise')
    # 9시대 포함, 10시대 제외. utztn_nope 합산이나 중복 제거는 하지 않는다.
    am_selected = am_selected.loc[am_times.dt.hour.ge(6) & am_times.dt.hour.lt(10)]
    am_counts[f'board_{am_year}'] = am_selected.groupby('셀 ID').size().reindex(am_counts.index,fill_value=0)
am_counts['change'] = am_counts['board_2025']-am_counts['board_2024']
am_reference = pd.read_csv(am_root/'results/analysis/feeder_bus/tables/H11a_격자별_공급구조_GTX역행.csv',encoding='utf-8-sig').set_index('셀 ID')
np.testing.assert_array_equal(am_counts['change'],am_reference.loc[am_counts.index,'Δ광역 출근'])

# 기존 원본과 같은 행정계 필터·중복 제거·생성 순서로 셀 ID를 재현한다.
am_admin_path = am_root/'data/raw/common/api_cache_gtx_a_routes/HangJeongDong_ver20260701.geojson'
am_grid_path = next((am_root/'data/raw/common/grid/빈격자(250m)').glob('*.shp'))
am_grid_crs = gpd.read_file(am_grid_path,rows=1).crs
am_admin = gpd.read_file(am_admin_path).to_crs(am_grid_crs)
am_admin['sgg_code'] = am_admin['sgg'].astype(str).str.extract(r'([0-9]+)')[0].str.zfill(5)
am_city = am_admin[am_admin['sgg_code'].str.startswith(('4128','4148'))]
am_boundary = am_city.geometry.union_all() if hasattr(am_city.geometry,'union_all') else am_city.geometry.unary_union
am_geometry = gpd.read_file(am_grid_path,bbox=am_boundary.bounds)
am_geometry = am_geometry[am_geometry.geometry.intersects(am_boundary)].copy()
am_geometry['_key'] = am_geometry.geometry.apply(lambda g:g.wkb if g is not None else None)
am_geometry = am_geometry.drop_duplicates('_key').drop(columns='_key').reset_index(drop=True)
am_geometry['셀 ID'] = np.arange(1,len(am_geometry)+1)
am_map_grid = am_geometry[['셀 ID','geometry']].merge(am_counts.reset_index(),on='셀 ID',how='inner',validate='one_to_one')
assert len(am_map_grid)==126 and am_map_grid['셀 ID'].is_unique
assert am_map_grid[['board_2024','board_2025','change']].notna().all().all()
am_map_edges = tuple(am_map_grid['change'].quantile([.25,.50,.75]))
am_map_grid['변화량 사분위 코드'] = np.searchsorted(am_map_edges,am_map_grid['change'].to_numpy(),side='left')
# 종일 페이지는 원래 S1 값으로 유지하고 출근 페이지와 별도 URL로 제공한다.
am_day_grid = am_geometry[['셀 ID','geometry']].merge(
    am_base[['셀 ID','nearest_station','board_2024','board_2025','change']],on='셀 ID',how='inner',validate='one_to_one')
am_day_edges = tuple(am_day_grid['change'].quantile([.25,.50,.75]))
am_day_grid['변화량 사분위 코드'] = np.searchsorted(am_day_edges,am_day_grid['change'].to_numpy(),side='left')
print('출근시간: 06:00~09:59:59 · 126개 격자 전체')
print('2024년:',int(am_map_grid['board_2024'].sum()),'건 → 2025년:',int(am_map_grid['board_2025'].sum()),'건')
print('변화량:',int(am_map_grid['change'].sum()),'건 · 사분위 경계:',am_map_edges)
print('감소:',int(am_map_grid['change'].lt(0).sum()),'동일:',int(am_map_grid['change'].eq(0).sum()),'증가:',int(am_map_grid['change'].gt(0).sum()))

# 지도 서버: 최신 모듈을 읽고 종일·출근 지도를 같은 포트에 각각 등록한다.
am_overlay_path = am_root/'code/03_eda/kakao_grid_overlay.py'
am_old_module = sys.modules.get('kakao_grid_overlay')
am_previous_pages = dict(getattr(am_old_module,'_pages',{})) if am_old_module is not None else {}
if am_old_module is not None and getattr(am_old_module,'_server',None) is not None:
    am_old_module._server.shutdown()
    am_old_module._server.server_close()
    if getattr(am_old_module,'_thread',None) is not None:
        am_old_module._thread.join(timeout=5)
am_overlay_module = types.ModuleType('kakao_grid_overlay')
am_overlay_module.__file__ = str(am_overlay_path)
exec(compile(am_overlay_path.read_text(encoding='utf-8'),str(am_overlay_path),'exec'),am_overlay_module.__dict__)
am_overlay_module._pages.update(am_previous_pages)
sys.modules['kakao_grid_overlay'] = am_overlay_module
am_station_coords = {'운정중앙':(37.716670,126.728330),'킨텍스':(37.665000,126.748060),'대곡':(37.631626,126.811024)}
am_kakao_key = os.getenv('KAKAO_JAVASCRIPT_KEY','').strip() or getpass.getpass('카카오 JavaScript 키: ').strip()
try:
    # 기존 종일 iframe을 유지하되, 이 셀에서는 종일 지도를 중복 표시하지 않는다.
    am_overlay_module.show_kakao_overlay(am_day_grid,am_station_coords,am_day_edges,am_kakao_key,
        count_basis=True,view_id='count',display_maps=False)
    am_overlay_module.show_kakao_overlay(am_map_grid,am_station_coords,am_map_edges,am_kakao_key,
        count_basis=True,view_id='morning_count',time_label='출근시간 06:00~09:59:59')
finally:
    del am_kakao_key
```

### 카카오맵: 하루 전체·출근시간대 통합 격자 지도

논문의 정적 지도(그림 6·9·12)와 같은 **한 장짜리 통합 지도**이다. **바로 아래 코드 셀 하나만 실행**하면 원자료 확인 후 JavaScript 키를 입력하고 운정중앙·킨텍스·대곡 지도를 볼 수 있다.

- 격자 **왼쪽 절반은 하루 전체**, **오른쪽 절반은 출근시간대(06:00~09:59)**이다.
- **감소 빨강 / 유지 회색 / 증가 파랑**, 두 시간대의 변화 구분이 다른 격자만 **격자 그림자**이다. 감소 TOP 3 표시와 사분위 구분은 사용하지 않는다.
- 운정중앙·대곡은 정적 지도와 같은 주요 지역에서 시작한다. 드래그로 자유롭게 이동하고 마우스 휠이나 확대·축소 버튼으로 범위를 조절할 수 있다. 작은 전체 지도와 범위 전환 버튼은 표시하지 않는다.
- 지도에는 건수 숫자를 표시하지 않는다. **격자를 클릭하면** 두 시간대의 2024·2025년 건수와 변화량을 확인할 수 있다.
- 각 지도 위의 **새 탭에서 크게 보기·캡처** 링크를 이용하면 논문에 넣을 화면을 직접 캡처하기 편하다.

기존과 같이 카카오 JavaScript SDK 도메인에 `http://localhost:8765`를 등록한다. 키는 실행 시 숨김 입력하거나 `KAKAO_JAVASCRIPT_KEY` 환경변수로 전달한다. 기존 종일·출근 지도 서버를 공유하므로 앞의 지도 셀을 먼저 실행할 필요는 없다. 키와 지도 HTML은 파일로 저장하지 않는다. [카카오 지도 API 공식 문서](https://apis.map.kakao.com/web/documentation/)

분석격자 주변 주거단지는 출근시간대 지도와 같은 반투명 연보라색으로 표시합니다. 시설·철도 자료도 기존에 저장한 자료를 재사용하며 격자 뒤에 배치합니다. 변화 구분이 다른 격자는 뒤에 부드러운 그림자를 넣어 살짝 떠 있는 것처럼 표시합니다. 격자 자체의 위치는 바꾸지 않습니다. 새 탭에서 C를 누르면 지도만 표시됩니다.


```python
# 정적 지도와 같은 통합 카카오맵: 이 셀만 실행 → JavaScript 키 입력
from pathlib import Path
import getpass
import os
import sys
import types
from IPython.display import display

combined_root = next((p for p in [Path.cwd(), *Path.cwd().parents]
                      if (p / 'config.py').is_file() and (p / 'code').is_dir()), None)
if combined_root is None:
    raise FileNotFoundError('UROP 프로젝트 또는 그 하위 폴더에서 실행해 주세요.')
combined_overlay_path = combined_root / 'code/03_eda/kakao_combined_grid_overlay.py'
combined_overlay = types.ModuleType('kakao_combined_grid_overlay')
combined_overlay.__file__ = str(combined_overlay_path)
exec(compile(combined_overlay_path.read_text(encoding='utf-8'), str(combined_overlay_path), 'exec'),
     combined_overlay.__dict__)
sys.modules['kakao_combined_grid_overlay'] = combined_overlay

combined_map_grid, combined_map_summary = combined_overlay.load_combined_grid(combined_root)
print('2024-10-17 / 2025-11-06 · 교통카드 행 수 · 126개 동일 격자')
print('격자 왼쪽=하루 전체 / 오른쪽=출근시간대 · 격자 그림자=변화 구분 불일치')
display(combined_map_summary)
combined_kakao_key = os.getenv('KAKAO_JAVASCRIPT_KEY', '').strip() or getpass.getpass('카카오 JavaScript 키: ').strip()
try:
    combined_map_links = combined_overlay.show_combined_kakao(
        combined_map_grid, combined_root, combined_kakao_key)
finally:
    del combined_kakao_key
```

## 3. 250m 격자별 GTX-A역과의 거리구간별 승차인원 변화

최근접 GTX-A역과 격자 중심점 사이의 직선거리를 비누적 구간으로 구분한다. 따라서 각 격자는 하나의 거리구간에만 포함된다.


```python
# 250m 격자별 GTX-A역과의 비누적 거리구간별 승차인원 변화
# 이 셀은 앞 셀 실행 여부와 관계없이 단독으로 실행할 수 있다.
from pathlib import Path
import geopandas as gpd
import numpy as np
import pandas as pd
from IPython.display import display

PROJECT_DISTANCE = UROP_ROOT
PROCESSED_DISTANCE = PROJECT_DISTANCE / 'data' / 'processed' / 'gtx_a'
FIGURES_DISTANCE = PROJECT_DISTANCE / 'results' / 'eda' / 'figures_011'
FIGURES_DISTANCE.mkdir(parents=True, exist_ok=True)

MODEL_DISTANCE_PATH = next(PROCESSED_DISTANCE.glob('18_*.csv'))
distance_model = pd.read_csv(MODEL_DISTANCE_PATH, encoding='utf-8-sig')
distance_model['셀 ID'] = pd.to_numeric(distance_model['셀 ID'], errors='coerce').astype('Int64')

# 격자 중심점과 GTX-A 3개 역을 EPSG:5179로 투영한 뒤 km 단위 직선거리를 계산한다.
distance_cells = gpd.GeoDataFrame(
    distance_model.copy(),
    geometry=gpd.points_from_xy(
        distance_model['중심점 경도'], distance_model['중심점 위도']
    ),
    crs='EPSG:4326',
).to_crs('EPSG:5179')

distance_stations = gpd.GeoDataFrame(
    {'GTX-A역': ['운정중앙역', '킨텍스역', '대곡역']},
    geometry=gpd.points_from_xy(
        [126.728330, 126.748060, 126.811024],
        [37.716670, 37.665000, 37.631626],
    ),
    crs='EPSG:4326',
).to_crs('EPSG:5179')

distance_matrix = np.column_stack([
    distance_cells.geometry.distance(station_geometry) / 1000
    for station_geometry in distance_stations.geometry
])
distance_cells['최근접 GTX-A역'] = np.asarray(distance_stations['GTX-A역'])[
    distance_matrix.argmin(axis=1)
]
distance_cells['최근접역 거리(km)'] = distance_matrix.min(axis=1)

# 0~1km부터 4~5km까지는 1km 간격이며, 5km 이상은 하나로 묶는다.
distance_bins = [0, 1, 2, 3, 4, 5, np.inf]
distance_labels = ['0~1km', '1~2km', '2~3km', '3~4km', '4~5km', '5km 이상']
distance_cells['GTX-A역 거리구간'] = pd.cut(
    distance_cells['최근접역 거리(km)'],
    bins=distance_bins,
    labels=distance_labels,
    right=False,
    include_lowest=True,
)
distance_cells['감소 셀 여부'] = distance_cells['change_all'] < 0

distance_table = (
    distance_cells.groupby('GTX-A역 거리구간', observed=False)
    .agg(
        셀_수=('셀 ID', 'size'),
        **{
            '2024년 승차인원': ('board_2024_all', 'sum'),
            '2025년 승차인원': ('board_2025_all', 'sum'),
            '승차인원 변화량': ('change_all', 'sum'),
            '승차인원 변화 중앙값': ('change_all', 'median'),
            '감소 셀': ('감소 셀 여부', 'sum'),
        },
    )
    .reset_index()
    .rename(columns={'셀_수': '셀 수'})
)
distance_table['승차인원 변화율(%)'] = np.where(
    distance_table['2024년 승차인원'] > 0,
    distance_table['승차인원 변화량'] / distance_table['2024년 승차인원'] * 100,
    np.nan,
)
distance_table['감소 비율(%)'] = np.where(
    distance_table['셀 수'] > 0,
    distance_table['감소 셀'] / distance_table['셀 수'] * 100,
    np.nan,
)

distance_table = distance_table[[
    'GTX-A역 거리구간', '셀 수', '승차인원 변화 중앙값',
    '감소 셀', '감소 비율(%)'
]]

total_row = pd.DataFrame([{
    'GTX-A역 거리구간': '전체',
    '셀 수': len(distance_cells),
    '승차인원 변화 중앙값': distance_cells['change_all'].median(),
    '감소 셀': int(distance_cells['감소 셀 여부'].sum()),
    '감소 비율(%)': distance_cells['감소 셀 여부'].mean() * 100,
}])
distance_table_with_total = pd.concat([distance_table, total_row], ignore_index=True)

print('[250m 격자별 GTX-A역과의 거리구간별 승차인원 변화]')
display(distance_table_with_total.round(1))

distance_output = FIGURES_DISTANCE / '04_GTX-A역_거리구간별_승차인원변화_요약.csv'
distance_table_with_total.to_csv(distance_output, index=False, encoding='utf-8-sig')
print('저장:', distance_output)
```

## 5. 노선별 승차 변화율

공급구조가 유지된 126개 셀에 포함된 승차기록만 노선별로 합산한다. 노선 수가 많으므로 변화율 순서대로 절반씩 나누어 두 개의 그림으로 제시한다.


```python
# 노선별 승차 변화율: 노선을 절반씩 나누어 두 개의 그림으로 출력
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from IPython.display import display

PROJECT_ROUTE = UROP_ROOT
ROOT_ROUTE = UROP_ROOT
PROCESSED_ROUTE = PROJECT_ROUTE / 'data' / 'processed' / 'gtx_a'
FIGURES_ROUTE = PROJECT_ROUTE / 'results' / 'eda' / 'figures_011'
FIGURES_ROUTE.mkdir(parents=True, exist_ok=True)

STABLE_ROUTE_PATH = next(PROCESSED_ROUTE.glob('16_*.csv'))
CARD_DIR_ROUTE = ROOT_ROUTE / 'data' / 'raw' / 'gtx_a' / 'transport_card'
RAW24_ROUTE = CARD_DIR_ROUTE / 'gtx_a_transport_card_20241017_raw_with_coords_api_filled.csv'
RAW25_ROUTE = CARD_DIR_ROUTE / 'gtx_a_transport_card_20251106_raw_with_coords.csv'

font_path_route = Path(r'C:\Windows\Fonts\malgun.ttf')
if font_path_route.exists():
    fm.fontManager.addfont(str(font_path_route))
    plt.rcParams['font.family'] = fm.FontProperties(fname=str(font_path_route)).get_name()
plt.rcParams['axes.unicode_minus'] = False

stable_route = pd.read_csv(STABLE_ROUTE_PATH, encoding='utf-8-sig')
stable_route['셀 ID'] = pd.to_numeric(stable_route['셀 ID'], errors='coerce').astype('Int64')

def split_stop_ids(value):
    return [
        item.strip() for item in str(value).split('|')
        if item.strip() and item.strip().lower() != 'nan'
    ]

stop_to_cell = {}
for year, column in [(2024, '2024년 정류장 ID'), (2025, '2025년 정류장 ID')]:
    stop_to_cell[year] = {
        stop_id: int(row['셀 ID'])
        for _, row in stable_route.iterrows()
        for stop_id in split_stop_ids(row[column])
    }

def aggregate_route_boarding(raw_path, year):
    data = pd.read_csv(raw_path, encoding='utf-8-sig', low_memory=False)
    stop_id = data['ride_sttn_id'].astype(str).str.replace(r'\.0$', '', regex=True)
    data['셀 ID'] = stop_id.map(stop_to_cell[year])
    data['승차인원'] = pd.to_numeric(data['utztn_nope'], errors='coerce').fillna(0)
    data['노선'] = data['query_route_no'].astype(str).str.replace(r'\.0$', '', regex=True)
    return data[data['셀 ID'].notna()].groupby('노선')['승차인원'].sum()

route_2024 = aggregate_route_boarding(RAW24_ROUTE, 2024)
route_2025 = aggregate_route_boarding(RAW25_ROUTE, 2025)
route_change = pd.concat([
    route_2024.rename('2024년 승차인원'),
    route_2025.rename('2025년 승차인원'),
], axis=1).fillna(0).reset_index()
route_change['증감'] = route_change['2025년 승차인원'] - route_change['2024년 승차인원']
route_change['변화율(%)'] = np.where(
    route_change['2024년 승차인원'] > 0,
    route_change['증감'] / route_change['2024년 승차인원'] * 100,
    np.nan,
)
route_change = route_change.sort_values('변화율(%)').reset_index(drop=True)
display(route_change.round(1))
# 두 그림 모두 전체 노선의 최솟값·최댓값을 이용한 동일한 x축 범위를 사용한다.
finite_rate = route_change['변화율(%)'].replace([np.inf, -np.inf], np.nan).dropna()
x_padding = max(5.0, float(finite_rate.max() - finite_rate.min()) * 0.04)
shared_x_min = float(finite_rate.min()) - x_padding
shared_x_max = float(finite_rate.max()) + x_padding

midpoint = int(np.ceil(len(route_change) / 2))
route_parts = [route_change.iloc[:midpoint], route_change.iloc[midpoint:]]

for figure_number, route_part in enumerate(route_parts, start=1):
    fig_height = max(5.8, len(route_part) * 0.52)
    fig, ax = plt.subplots(figsize=(11.5, fig_height))
    colors = np.where(route_part['변화율(%)'] < 0, '#2166ac', '#d73027')
    ax.barh(route_part['노선'], route_part['변화율(%)'], color=colors, alpha=0.9)
    ax.axvline(0, color='black', linewidth=1)
    ax.set_xlim(shared_x_min, shared_x_max)
    ax.set_xlabel('승차인원 변화율(%)')
    ax.set_ylabel('노선번호', labelpad=12)
    ax.set_title(f'노선별 승차인원 변화율 ({figure_number}/2)')
    ax.spines[['top', 'right']].set_visible(False)
    ax.grid(axis='x', alpha=0.2)
    ax.tick_params(axis='y', pad=8, labelsize=10)

    # 변화율 수치는 막대 끝이 아닌 오른쪽 고정 열에 배치하여 노선번호와 겹치지 않게 한다.
    for y_position, value in enumerate(route_part['변화율(%)']):
        if pd.notna(value):
            ax.text(
                1.025, y_position, f'{value:.1f}%',
                transform=ax.get_yaxis_transform(),
                va='center', ha='left', fontsize=9, clip_on=False,
                bbox=dict(facecolor='white', edgecolor='none', alpha=0.78, pad=1.0),
            )

    # 오른쪽 여백은 변화율 숫자만 표시하는 별도 공간이다.
    fig.subplots_adjust(left=0.16, right=0.83, top=0.90, bottom=0.12)
    plt.show()
```

### 5-1. 노선별 승차 변화율 — 승차건수 기준

## 7. A3 Ridge Logistic OOF 오차행렬

12-5의 중첩 Leave-One-Station-Out OOF 예측 결과를 전체 및 역 권역별 오차행렬로 표시한다. 분류 기준은 증가확률 0.5이며, 괄호 안 비율은 각 실제 집단 행 안에서의 비율이다.

```python
# A3 중첩 LOSO OOF 오차행렬 시각화
# 12-5에서 검증된 OOF 집계값을 사용한다.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

plt.rcParams['font.family'] = 'Apple SD Gothic Neo'
plt.rcParams['axes.unicode_minus'] = False

confusion_results = {
    '전체 OOF': np.array([[26, 34], [17, 49]]),
    '대곡': np.array([[4, 15], [2, 16]]),
    '운정중앙': np.array([[11, 18], [4, 30]]),
    '킨텍스': np.array([[11, 1], [11, 3]]),
}

def annotations(cm):
    row_total = cm.sum(axis=1, keepdims=True)
    row_pct = np.divide(cm, row_total, out=np.zeros_like(cm, dtype=float), where=row_total != 0) * 100
    return np.array([
        [f'{cm[i, j]}개\n({row_pct[i, j]:.1f}%)' for j in range(2)]
        for i in range(2)
    ])

def balanced_accuracy_from_cm(cm):
    specificity = cm[0, 0] / cm[0].sum()
    sensitivity = cm[1, 1] / cm[1].sum()
    return (specificity + sensitivity) / 2

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
for ax, (name, cm) in zip(axes.flat, confusion_results.items()):
    sns.heatmap(
        cm, annot=annotations(cm), fmt='', cmap='Blues', cbar=False,
        linewidths=1.5, linecolor='white', square=True, ax=ax,
        xticklabels=['비확대 예측(0)', '확대 예측(1)'],
        yticklabels=['실제 비확대(0)', '실제 확대(1)'],
        annot_kws={'size': 13, 'weight': 'bold'},
    )
    n = int(cm.sum())
    balanced_accuracy = balanced_accuracy_from_cm(cm)
    ax.set_title(
        f'{name} (n={n}, Balanced Accuracy={balanced_accuracy:.3f})',
        fontsize=13, weight='bold'
    )
    ax.set_xlabel('예측값')
    ax.set_ylabel('실제값')
    ax.tick_params(axis='x', rotation=0)
    ax.tick_params(axis='y', rotation=0)

fig.suptitle(
    'A3 Ridge Logistic 중첩 LOSO OOF 오차행렬 (기준확률 0.5)',
    fontsize=16, weight='bold', y=1.02
)
plt.tight_layout()
plt.show()

metric_rows = []
for name, cm in confusion_results.items():
    tn, fp, fn, tp = cm.ravel()
    metric_rows.append({
        '구분': name, '표본 수': cm.sum(),
        'TN': tn, 'FP': fp, 'FN': fn, 'TP': tp,
        'Accuracy': (tn + tp) / cm.sum(),
        'Balanced Accuracy': balanced_accuracy_from_cm(cm),
        'Sensitivity': tp / (tp + fn),
        'Specificity': tn / (tn + fp),
    })
metrics = pd.DataFrame(metric_rows)
display(metrics.round(3))
```

- 실제 비확대 격자 60개 중 26개만 비확대로 맞혀 특이도는 상대적으로 낮다.

- ROC-AUC는 임계값 전체에서 순위 판별력을 평가하고, 이 오차행렬은 임계값 0.5에서의 최종 분류 결과를 보여준다.

## 8. 쌍대 비교 Ridge 순위모형 성능 시각화

같은 GTX역 권역에서 실제 수요 비중 확대 격자가 비확대 격자보다 높은 점수를 받도록 학습한 순위 전용 모형의 외부 검증 결과이다. 각 외부 검증에서 해당 역 권역을 학습과 규제강도 선택에서 완전히 제외한 중첩 Leave-One-Station-Out OOF 점수를 사용한다.

순위모형은 확대확률이나 0.5 분류기준을 사용하지 않으므로 오차행렬보다 다음 지표가 목적에 직접 대응한다.

- **ROC-AUC**: 실제 확대 격자를 비확대 격자보다 위에 배치하는 능력
- **Pairwise Log Loss**: 올바른 순서에 부여한 확신의 적절성. 낮을수록 좋으며 정보 없는 기준은 0.6931
- **Precision@K**: 예측점수 상위 K% 중 실제 확대 격자의 비율
- **Lift@K**: Precision@K를 해당 권역 전체 확대 비율로 나눈 값


```python
# 쌍대 비교 Ridge 순위모형: 중첩 LOSO 성능 시각화
from pathlib import Path
import importlib.util
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns
from sklearn.metrics import roc_auc_score, roc_curve
from IPython.display import display

# 노트북을 저장소 루트 또는 analysis 폴더에서 실행하는 경우 모두 지원
_candidates = [
    *(UROP_ROOT / 'code').rglob('12-8_A3_쌍대비교_순위모형.py'),
    Path.cwd() / '12-8_A3_쌍대비교_순위모형.py',
    Path.cwd() / 'gtx_a_bus_analysis/analysis/12-8_A3_쌍대비교_순위모형.py',
]
RANK_SCRIPT = next((p for p in _candidates if p.exists()), None)
if RANK_SCRIPT is None:
    raise FileNotFoundError('12-8_A3_쌍대비교_순위모형.py를 찾을 수 없습니다.')

spec = importlib.util.spec_from_file_location('pairwise_rank_viz', RANK_SCRIPT)
rank = importlib.util.module_from_spec(spec)
spec.loader.exec_module(rank)
base = rank.load_base_module()
rank_data = base.load_data().dropna(
    subset=['nearest_station', 'share_increase', *rank.FEATURES]
).copy()
rank_data['share_increase'] = rank_data['share_increase'].astype(int)

oof_rows, fold_rows, top_rows = [], [], []
all_pair_losses = []
for test_station in rank_data['nearest_station'].astype(str).unique():
    train = rank_data[rank_data['nearest_station'].astype(str) != test_station]
    test = rank_data[rank_data['nearest_station'].astype(str) == test_station].copy()
    best_c, inner_auc = rank.choose_c_inner_loso(train)
    pair_x, pair_y = rank.make_pairs(train)
    fitted_ranker = rank.make_model(best_c).fit(pair_x, pair_y)
    test['ranking_score'] = rank.cell_scores(fitted_ranker, test)

    pair_loss, pair_count = rank.pairwise_log_loss(test, test['ranking_score'].to_numpy())
    pos = test.loc[test['share_increase'] == 1, 'ranking_score'].to_numpy()
    neg = test.loc[test['share_increase'] == 0, 'ranking_score'].to_numpy()
    all_pair_losses.append(np.logaddexp(0.0, -(pos[:, None] - neg[None, :]).ravel()))
    fold_rows.append({
        '검증역': test_station,
        '검증 격자 수': len(test),
        '선택 C': best_c,
        'ROC-AUC': roc_auc_score(test['share_increase'], test['ranking_score']),
        'Pairwise Log Loss': pair_loss,
        '비교쌍 수': pair_count,
    })
    top_rows.extend(rank.precision_lift_rows(test, test['ranking_score'].to_numpy()))
    oof_rows.append(test[['nearest_station', 'share_increase', 'ranking_score']])

rank_oof = pd.concat(oof_rows, ignore_index=True)
rank_fold = pd.DataFrame(fold_rows)
rank_top = pd.DataFrame(top_rows)
rank_top['상위 비율(%)'] = rank_top['상위 구간'].str.extract(r'(\d+)').astype(int)

# 한글 글꼴
for _font in ['AppleGothic', 'Malgun Gothic', 'NanumGothic']:
    if any(f.name == _font for f in fm.fontManager.ttflist):
        plt.rcParams['font.family'] = _font
        break
plt.rcParams['axes.unicode_minus'] = False
sns.set_theme(style='whitegrid', font=plt.rcParams.get('font.family', ['sans-serif'])[0])

station_order = rank_fold['검증역'].tolist()
colors = dict(zip(station_order, sns.color_palette('Set2', len(station_order))))
fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# (a) 역별 ROC 곡선
ax = axes[0, 0]
for station in station_order:
    part = rank_oof[rank_oof['nearest_station'].astype(str) == station]
    fpr, tpr, _ = roc_curve(part['share_increase'], part['ranking_score'])
    auc = roc_auc_score(part['share_increase'], part['ranking_score'])
    ax.plot(fpr, tpr, lw=2.4, color=colors[station], label=f'{station} (AUC={auc:.3f})')
ax.plot([0, 1], [0, 1], '--', color='gray', lw=1.5, label='무작위 기준 (0.5)')
ax.set(xlabel='False Positive Rate', ylabel='True Positive Rate', title='(a) 외부 검증역별 ROC 곡선')
ax.legend(loc='lower right')

# (b) Pairwise Log Loss
ax = axes[0, 1]
bars = ax.bar(rank_fold['검증역'], rank_fold['Pairwise Log Loss'],
              color=[colors[s] for s in rank_fold['검증역']], edgecolor='black', linewidth=.6)
ax.axhline(np.log(2), color='crimson', linestyle='--', lw=2, label='정보 없음 기준 0.6931')
for bar, value in zip(bars, rank_fold['Pairwise Log Loss']):
    ax.text(bar.get_x() + bar.get_width()/2, value + .012, f'{value:.3f}', ha='center', fontsize=11)
ax.set_ylim(0, 0.78)
ax.set(ylabel='Pairwise Log Loss (낮을수록 좋음)', title='(b) 외부 검증역별 순위 손실')
ax.legend()

# (c) Precision@K
ax = axes[1, 0]
for station in station_order:
    part = rank_top[rank_top['검증역'].astype(str) == station].sort_values('상위 비율(%)')
    ax.plot(part['상위 비율(%)'], part['Precision@K'], marker='o', ms=7, lw=2.3,
            color=colors[station], label=station)
    baseline = part['권역 실제 확대 비율'].iloc[0]
    ax.hlines(baseline, 9, 31, colors=[colors[station]], linestyles=':', alpha=.75)
ax.set_xticks([10, 20, 30])
ax.set_ylim(0.35, 1.05)
ax.set(xlabel='예측점수 상위 구간(%)', ylabel='실제 확대 격자 비율', title='(c) Precision@K (점선: 권역 전체 확대 비율)')
ax.legend()

# (d) Lift@K
ax = axes[1, 1]
for station in station_order:
    part = rank_top[rank_top['검증역'].astype(str) == station].sort_values('상위 비율(%)')
    ax.plot(part['상위 비율(%)'], part['Lift'], marker='o', ms=7, lw=2.3,
            color=colors[station], label=station)
ax.axhline(1.0, color='gray', linestyle='--', lw=1.5, label='무작위 선정 기준')
ax.set_xticks([10, 20, 30])
ax.set_ylim(0.9, 2.0)
ax.set(xlabel='예측점수 상위 구간(%)', ylabel='Lift', title='(d) Lift@K (1보다 클수록 상위 집중)')
ax.legend()

fig.suptitle('쌍대 비교 Ridge 순위모형의 중첩 LOSO 외부검증', fontsize=17, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

rank_summary = rank_fold.copy()
rank_summary.loc[len(rank_summary)] = {
    '검증역': '요약',
    '검증 격자 수': rank_summary['검증 격자 수'].sum(),
    '선택 C': np.nan,
    'ROC-AUC': rank_summary['ROC-AUC'].mean(),
    'Pairwise Log Loss': np.concatenate(all_pair_losses).mean(),
    '비교쌍 수': rank_summary['비교쌍 수'].sum(),
}
display(rank_summary.round(4))
display(rank_top[['검증역', '상위 구간', 'K', '권역 실제 확대 비율', 'Precision@K', 'Lift']].round(4))
```

- 전체 외부 비교쌍의 Pairwise Log Loss는 0.539로, 모든 쌍에 50%를 부여하는 정보 없는 기준 0.693보다 낮다.

## GTX역 연계버스 승차건수 도넛

운정중앙 핑크·킨텍스 회색·대곡 갈색. 도넛에는 건수·비중을, 아래에는 권역별 색상 범례를 표시합니다. 파일 저장 없이 출력에서 복사하세요.


```python
# 하루 전체 연계버스 도넛: 건수·비중 + 권역 범례 · 이미지 파일 저장 없음
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from IPython.display import display
from matplotlib.patches import Patch

donut_root = next(p for p in [Path.cwd(), *Path.cwd().parents]
                  if (p / 'config.py').is_file() and (p / 'code').is_dir())
donut_data = pd.read_csv(
    donut_root / 'results/analysis/feeder_bus/tables/H11a_격자별_공급구조_GTX역행.csv',
    encoding='utf-8-sig')
donut_base = pd.read_csv(donut_root / 'data/final/gtx_a/S1_126격자_건수기준_기초표.csv',
                         encoding='utf-8-sig')
assert len(donut_data) == 126 and donut_data['셀 ID'].is_unique
assert set(donut_data['셀 ID']) == set(donut_base['셀 ID'])
pd.testing.assert_series_equal(
    donut_data.set_index('셀 ID')['권역'].sort_index(),
    donut_base.set_index('셀 ID')['nearest_station'].sort_index(), check_names=False)
donut_order = ['운정중앙', '킨텍스', '대곡']
donut_values = donut_data.groupby('권역')['GTX역행 2025'].sum().reindex(donut_order)
assert donut_values.notna().all() and donut_values.ge(0).all()
np.testing.assert_array_equal(donut_values, [672, 113, 230])
donut_total = int(donut_values.sum())
assert donut_total == 1015
donut_colors = ['#C45A88', '#888888', '#8A6446']
donut_font_path = next((p for p in [
    Path.home() / 'Library/Fonts/KoPubWorld Dotum Bold.ttf',
    Path('/Library/Fonts/NanumGothicBold.ttf'),
    Path('/System/Library/Fonts/AppleSDGothicNeo.ttc')
] if p.exists()), None)
if donut_font_path is not None:
    fm.fontManager.addfont(str(donut_font_path))
donut_font_name = (fm.FontProperties(fname=str(donut_font_path)).get_name()
                   if donut_font_path is not None else 'DejaVu Sans')

with plt.rc_context({'font.family': donut_font_name, 'font.weight': 'bold'}):
    donut_fig, donut_ax = plt.subplots(figsize=(7.6, 8.2), facecolor='none')
    donut_fig.subplots_adjust(left=.02, right=.98, bottom=.12, top=.98)
    donut_ax.pie(
        donut_values.to_numpy(), colors=donut_colors, startangle=90, counterclock=False,
        wedgeprops={'width': .44, 'edgecolor': 'white', 'linewidth': 3},
        autopct=lambda p: f'{round(p * donut_total / 100):,}건\n{p:.1f}%',
        pctdistance=.78,
        textprops={'fontsize': 20, 'fontweight': 'bold', 'color': 'white',
                   'linespacing': 1.35})
    donut_ax.text(0, 0, f'{donut_total:,}건', ha='center', va='center',
                  fontsize=34, fontweight='bold', color='#111111')
    donut_fig.legend(
        handles=[Patch(facecolor=color, edgecolor='none', label=station + '역')
                 for station, color in zip(donut_order, donut_colors)],
        loc='lower center', bbox_to_anchor=(.5, .025), ncol=3,
        frameon=False, fontsize=17, handlelength=1.15, handleheight=1.15,
        handletextpad=.5, columnspacing=1.3)
    donut_ax.set_aspect('equal')
    donut_ax.set_facecolor('none')
    display(donut_fig)
    plt.close(donut_fig)
```
